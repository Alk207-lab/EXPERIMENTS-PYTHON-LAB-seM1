Q1-Given an integer n find whether the number is even or odd. If the number is even, check if it is between 2 and 5, check if it is between 6 and 20, or greater than 20.

### Explanation for Q1: Conditional Logic (If-Elif-Else) and Modulo Operator

This code demonstrates the use of conditional statements (`if`, `elif`, `else`) to control the flow of a program based on certain conditions. It checks an integer `n` for specific properties:

*   **Modulo Operator (`%`):** `n % 2 != 0` checks if `n` is an odd number. The modulo operator returns the remainder of a division. If `n` divided by 2 has a non-zero remainder, it's odd.
*   **Chained Comparisons:** `2 <= n <= 5` is a convenient way in Python to check if a number falls within a specific range.
*   **Problem Logic:** The problem defines rules for classifying numbers as 'Weird' or 'Not Weird' based on whether they are odd or even and fall within certain ranges. The `if-elif-else` structure directly implements these rules, evaluating conditions in order and executing the first block whose condition is true.

```python
n = int(input("Enter an integer: "))

if n % 2 != 0:
    print("Weird")
elif 2 <= n <= 5:
    print("Not Weird")
elif 6 <= n <= 20:
    print("Weird")
else:  # n is even and n > 20
    print("Not Weird")
```

Q2. WAP to read an integer 'n' from STDIN. For all non-negative integers i<n, print i² on a separate line.


### Explanation Q2: Loops (`for` loop) and Exponentiation

This code introduces the concept of **iteration** using a `for` loop to perform a repetitive task. Specifically:

*   **`for i in range(n):`**: This loop iterates `n` times. The `range(n)` function generates a sequence of numbers from `0` up to (but not including) `n`. In each iteration, the variable `i` takes on the next value from this sequence.
*   **Exponentiation (`**`):** `i**2` calculates the square of the current value of `i`. The `**` operator in Python is used for exponentiation (raising a number to a power).
*   **Output:** `print(i**2)` prints the result of `i` squared on a new line for each iteration, fulfilling the requirement to print each square on a separate line.

```python
n= int(input("enter a number"))
for i in range(n):
  print(i**2)
```

Q3-Q3 WAP to read an integer from STDIN. Without using any string methods, print the following on a single line.
123....n


### Concept for Q3: Loops (`for` loop) and Controlling Print Output

This code again utilizes a `for` loop but focuses on how to control the output of the `print()` function to achieve a specific formatting requirement:

*   **`for i in range(1, n+1):`**: This loop iterates from `1` up to and including `n`. `range(start, stop)` generates numbers from `start` up to (but not including) `stop`. So, `n+1` is used to include `n` itself.
*   **`print(i, end="")`**: This is the key part for controlling output. By default, `print()` adds a newline character (`\n`) at the end of its output. The `end=""` argument explicitly tells the `print()` function to use an empty string as the ending character instead of a newline. This causes all the numbers printed in the loop to appear consecutively on the *same line* without any spaces in between, fulfilling the requirement `123....n`.

```python
n = int(input("enter the number"))
print("output is", end=" ")

for i in range(1, n+1):
    print(i, end="")
```
